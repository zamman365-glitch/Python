
# Q4. Employee and Developer Class

## Problem Statement
Create parent class `Employee` and child class `Developer`.

Store:
- Employee Name
- Salary
- Programming Language

Display all details.

## Sample Output
```python
Name: Aman
Salary: 70000
Language: Python
```


````python
# Write your code here
class Employee:
    def __init__(self, name, salary):
        self.name = name
        self.salary = salary

class Developer(Employee):
    def __init__(self, name, salary, language):
        super().__init__(name, salary)
        self.language = language

    def display_details(self):
        print(f"Name: {self.name}")
        print(f"Salary: {self.salary}")
        print(f"Language: {self.language}")
dev = Developer("Aman", 70000, "Python")
dev.display_details()
````


# Q5. Mobile and AndroidMobile Class

## Problem Statement
Create parent class `Mobile` and child class `AndroidMobile`.

Store:
- Brand
- RAM

Display details using method.

## Sample Output
```python
Brand: Samsung
RAM: 8GB
```


````python
# Write your code here
class Mobile:
    def __init__(self,Brand):
        self.Brand=Brand
       
class AndroidMobile(Mobile):
    def __init__(self,Brand,Ram):
        super().__init__(Brand)
        self.Ram=Ram

    def display(self):
        print(f"Brand:{self.Brand}")
        print(f"Ram:{self.Ram}")
    
obj=AndroidMobile("Samsung","8GB")
obj.display()
````


# Q6. Shape and Square Class

## Problem Statement
Create parent class `Shape` and child class `Square`.

Store:
- Side

Create method to calculate area.

## Formula
```python
Area = side * side
```

## Sample Output
```python
Area: 64
```


````python
# Write your code here
class Shape:
    def __init__(self,r):
        self.r=r

class Square(Shape):
    def display1(self):
        area=self.r*self.r
        print("Area:",area)
obj=Square(2)
obj.display1()
````
