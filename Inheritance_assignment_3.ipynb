# Q1. Electronics and Laptop Class

## Problem Statement
Create a parent class `Electronics` and child class `Laptop`.

Store:
- brand_name
- processor

Display all details using method.

## Sample Input
```python
Brand: Dell
Processor: i7
```

## Sample Output
```python
Brand: Dell
Processor: i7
```


````python
# Write your code here
class Electronics:
    def __init__(self,brand_name,processor):
        self.brand_name=brand_name
        self.processor=processor
class laptop(Electronics):
    def __init__(self,brand_name,processor):
        super().__init__(brand_name,processor)
    def display(self):
        print("Brand:",self.brand_name)
        print("Processor:",self.processor)
obj=laptop("Dell","i7")
obj.display()
````


# Q3. Country, State and City Class

## Problem Statement
Create multilevel inheritance using:
- Country
- State
- City

Store:
- country_name
- state_name
- city_name

Display all details.

## Sample Input
```python
Country: India
State: Madhya Pradesh
City: Bhopal
```

## Sample Output
```python
Country: India
State: Madhya Pradesh
City: Bhopal
```


````python
# Write your code here
class Country:
    def __init__(self,country_name):
        self.country_name=country_name
class State(Country):
    def __init__(self,country_name,state_name):
        super().__init__(country_name)
        self.state_name=state_name
class City(State):
    def __init__(self,country_name,state_name,city_name):
        super().__init__(country_name,state_name)
        self.city_name=city_name
    
    def display(self):
        print("Country:",self.country_name)
        print("State:",self.state_name)
        print("City:",self.city_name)

obj=City("India","MP","Bhopal")
obj.display()
````


# Q5. Father and Mother Multiple Inheritance

## Problem Statement
Create:
- Father class
- Mother class
- Son class

Store:
- father_profession
- mother_profession
- son_name

Display all details.

## Sample Input
```python
Father Profession: Engineer
Mother Profession: Teacher
Son Name: Rahul
```

## Sample Output
```python
Father Profession: Engineer
Mother Profession: Teacher
Son Name: Rahul
```


````python
# Write your code here
class Father:
    def __init__(self,father_profession):
        self.father_profession=father_profession
class Mother:
     def __init__(self,mother_profession):
        self.mother_profession=mother_profession
class Son(Father,Mother):
    def __init__(self,father_profession,mother_profession,son_name):
        self.father_profession=father_profession
        self.mother_profession=mother_profession
        self.son_name=son_name
    
    def display(self):
        print("Father profession:",self.father_profession)
        print("Mother Profession:",self.mother_profession)
        print("Son Name:",self.son_name)

obj=Son("Engineer",'Teacher','Rahul')
obj.display()

````


# Q6. Appliance, WashingMachine and Refrigerator Class

## Problem Statement
Create hierarchical inheritance using:
- Appliance
- WashingMachine
- Refrigerator

Store:
- brand_name
- capacity

Display all details.

## Sample Input
```python
Brand: LG
Capacity: 7KG
```

## Sample Output
```python
Washing Machine Brand: LG
Capacity: 7KG

Refrigerator Brand: Samsung
Capacity: 250L
```


````python
# Write your code here
class Appliance:
    def __init__(self,brand_name,capacity):
        self.brand_name=brand_name
        self.capacity=capacity

class WashingMachine(Appliance):
    def display1(self):
        print("Washing Machine Brand:",self.brand_name)
        print("Capacity:",self.capacity)

class refrigerator(Appliance):
    def display2(self):
        print("Refrigerator Brand:",self.brand_name)
        print("Capacity:",self.capacity)


obj1=WashingMachine("LG","7KG")
obj2=refrigerator("Samsung","250L")

obj1.display1()
print()
obj2.display2()
````
