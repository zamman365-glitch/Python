# Q1. Single Inheritance — Animal and Cat

## Problem Statement
Create a parent class `Animal` with:
- animal_name

Create child class `Cat` with:
- color

Create method to display all details.

## Sample Output
```python
Animal Name: Cat
Color: White
```


````python
# Write your code here
class Animal:
    def display1(self):
        print("Animal Name: Cat")
class Cat(Animal):
    def display2(self):
        print("Color: White")
obj=Cat()
obj.display1()
obj.display2()
````


# Q3. Multilevel Inheritance — Grandfather, Father, Son

## Problem Statement
Create:
- Grandfather class
- Father class inherited from Grandfather
- Son class inherited from Father

Store:
- grandfather_name
- father_name
- son_name

Display all details.

## Sample Output
```python
Grandfather: Mohan
Father: Rajesh
Son: Aman
```


````python
# Write your code here
class grandparents:
    def display1(self):
        print("GrandFather: Mohan")
class Father(grandparents):
    def display2(self):
        print("Father: Rajesh")
class Son(Father):
    def display3(self):
        print("Son: Aman")
obj=Son()
obj.display1()
obj.display2()
obj.display3()
````


# Q4. Multilevel Inheritance — School, Teacher, Student

## Problem Statement
Create:
- School class
- Teacher class inherited from School
- Student class inherited from Teacher

Store:
- school_name
- teacher_name
- student_name

Display all details.

## Sample Output
```python
School: DPS
Teacher: Neha
Student: Arjun
```


````python
# Write your code here
class School:
    def __init__(self,school_name):
        self.school_name=school_name

class Teacher(School):
    def __init__(self,school_name,teacher_name):
        super().__init__(school_name)
        self.teacher_name=teacher_name

class Student(Teacher):
    def __init__(self,school_name,teacher_name,student_name):
        super().__init__(school_name,teacher_name)
        self.student_name=student_name

    def display(self):
        print("School:",self.school_name)
        print("Teacher:",self.teacher_name)
        print("Student:",self.student_name)

obj=Student("DPS","Neha",'Arjun')
obj.display()
````


# Q5. Multiple Inheritance — Father and Mother

## Problem Statement
Create:
- Father class
- Mother class

Create child class `Child` inherited from both classes.

Store:
- father_name
- mother_name
- child_name

Display all details.

## Sample Output
```python
Father: Raj
Mother: Sunita
Child: Ravi
```


````python
# Write your code here
class father:
    def __init__(self,father_name):
        self.father_name=father_name

class mother:
    def __init__(self,mother_name):
        self.mother_name=mother_name

class Child(father,mother):
    def __init__(self,father_name,mother_name,son_name):
        self.father_name=father_name
        self.mother_name=mother_name
        self.son_name=son_name
    
    def display(self):
        print("Father:",self.father_name)
        print("Mother:",self.mother_name)
        print("Son:",self.son_name)

obj=Child("Raj",'Sunita',"Ravi")
obj.display()
````


# Q7. Hierarchical Inheritance — Vehicle, Car, Bike

## Problem Statement
Create parent class `Vehicle`.

Create two child classes:
- Car
- Bike

Store:
- company_name
- model_name

Display details using both child classes.

## Sample Output
```python
Car Company: BMW
Car Model: X5

Bike Company: Yamaha
Bike Model: R15
```


````python
# Write your code here
class Vehicle:
    def __init__(self, company_name, model_name):
        self.company_name = company_name
        self.model_name = model_name

class Car(Vehicle):
    def display_car(self):
        print("Car Company:", self.company_name)
        print("Car Model:", self.model_name)

class Bike(Vehicle):
    def display_bike(self):
        print("Bike Company:", self.company_name)
        print("Bike Model:", self.model_name)

obj1 = Car("BMW", "X5")
obj2 = Bike("Yamaha", "R15")

obj1.display_car()
print()
obj2.display_bike()
````


# Q8. Hierarchical Inheritance — Employee, Developer, Designer

## Problem Statement
Create parent class `Employee`.

Create two child classes:
- Developer
- Designer

Store:
- employee_name
- department

Display details.

## Sample Output
```python
Developer: Rahul
Department: Python

Designer: Neha
Department: UI/UX
```


````python
# Write your code here
class Employee:
    def __init__(self, employee_name, department):
        self.employee_name = employee_name
        self.department = department


class Developer(Employee):
    def display_developer(self):
        print("Developer:", self.employee_name)
        print("Department:", self.department)

class Designer(Employee):
    def display_designer(self):
        print("Designer:", self.employee_name)
        print("Department:", self.department)

obj1 = Developer("Rahul", "Python")
obj2 = Designer("Neha", "UI/UX")


obj1.display_developer()
print()
obj2.display_designer()
````
